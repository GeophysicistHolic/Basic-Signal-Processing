```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import BoundaryNorm
from matplotlib.ticker import MaxNLocator

lt = 128
dt = 1
t = np.arange(0, lt-1, dt)


x1 = np.cos(2*np.pi*10*(t/lt))
x2 = np.cos(2*np.pi*(10+(t/7))*(t/lt))
two_series = x1 +x2

H = np.fft.fft(two_series)


f = np.fft.fftfreq(len(t))
f = np.fft.fftshift(f)


h = np.zeros(len(t), dtype='complex')

for i in range(len(h)):
        h[i] = H[i]

M = np.zeros((len(h), len(f)))

for i in range(1,len(f)):
    G = np.exp((-2*(np.pi**2)*(f**2))/((f[i])**2))
    for j in range(len(h)):
        c = h[126]
        if j < int(len(h)-1):
            h[j] = h[j+1]
        h[126] = c
    a = G*h
    a = np.fft.ifft(a)
    M[:][i] = abs(a)


fig0, axs0 = plt.subplots(2, 1, sharex=True, constrained_layout=True)

axs0[0].plot(t, two_series)
axs0[0].set_title('Original Signal Two Series')
axs0[0].set(xlabel='Time(ms)', ylabel='Amplitude')

pcm = axs0[1].pcolormesh(t, abs(f[:64]), M[:][:64])
axs0[1].set_title('Gamma : 1, m : 1')
axs0[1].set(xlabel='Time(ms)', ylabel='Frekuensi (Hz)')

fig0.colorbar(pcm, ax=axs0[1], orientation='horizontal')
```
